```python
# Using TypeDict --> sugesstion not mendetory
import os
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from typing import TypedDict , Annotated , List , Optional
from datetime import datetime

load_dotenv()
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")

# Create Schema
class Itenary(TypedDict):
    key_themes : Annotated[List[str] , "give me the theme that we are flowwing for this day in details"]
    day : Annotated[int , "a what number of day it is"]
    time : Annotated[datetime , "a time frame gap it is, e.g (8:00:00 - 2:20:20)"]
    description : str
    estimated_amount : Annotated[Optional[float] , "Give me a esctimated amt of money for this in US $"]
    
llm = ChatOpenAI(model="openai/gpt-4o-mini", openai_api_base="https://openrouter.ai/api/v1", api_key=OPENROUTER_API_KEY) ## LLM 
structured_llm = llm.with_structured_output(Itenary)
results = structured_llm.invoke("give me itenary to travel in india for for 7 days day wise")
print(results)
```

```python
# Using pydentic --> give you proper validation inbuild
from pydantic import BaseModel , Field
from typing import Dict , List , Optional , Annotated

class ItenaryPydantic(BaseModel):
    key_themes : List[str] = Field(description= "give me the theme that we are flowwing for this day in details")
    day : Annotated[int , "a what number of day it is"]
    time : Annotated[datetime , "a time frame gap it is, e.g (8:00:00 - 2:20:20)"]
    description : str
    estimated_amount : Annotated[Optional[float] , "Give me a esctimated amt of money for this in indian ruppess"]
    
structured_pydantic_llm = llm.with_structured_output(schema=ItenaryPydantic)
results = structured_pydantic_llm.invoke("give me itenary to Delhi in delhi for for 1 days day wise")
print(results)
```
